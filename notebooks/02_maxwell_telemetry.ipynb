# MaxWell HD-MEA Telemetry
This notebook demonstrates how to load and visualize high-density electrophysiology data from MaxWell Biosystems.

```python
import matplotlib.pyplot as plt
from src.data.ephys.maxwell_dataset import MaxWellHDMEADataset

# Instantiate the MaxWellHDMEADataset
try:
    dataset = MaxWellHDMEADataset('data/ephys/sample.raw.h5', sequence_length=5000, target_channels=10)
    print(f"Loaded dataset with {len(dataset)} sequence chunks.")
    
    # Get the first temporal chunk
    chunk = dataset[0]
    
    # Visualize a slice of the telemetry
    plt.figure(figsize=(15, 8))
    time_frames = range(chunk.shape[0])
    
    # Plot as a stacked line plot
    for i in range(chunk.shape[1]):
        # Offset each channel to visualize them stacked
        offset = i * 5  # Adjust offset as necessary depending on the Z-scored signal amplitude
        plt.plot(time_frames, chunk[:, i] + offset, label=f'Channel {i+1}', linewidth=0.8)
        
    plt.title('MaxWell HD-MEA Telemetry - 10 Channels over 5000 Frames')
    plt.xlabel('Time Frame')
    plt.ylabel('Amplitude (Z-score + Offset)')
    plt.legend(loc='upper right', bbox_to_anchor=(1.15, 1))
    plt.tight_layout()
    plt.show()
    
except FileNotFoundError:
    print("\n--- File Not Found ---")
    print("The sample .raw.h5 file was not found at 'data/ephys/sample.raw.h5'.")
    print("Please download the MaxWell sample dataset from their documentation portal")
    print("or place your own .raw.h5 recording file into the data/ephys/ directory.")
```
