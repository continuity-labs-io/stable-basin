# Müller-Brown Toy Physics
This notebook demonstrates the Müller-Brown potential and generated Langevin trajectories.

```python
import numpy as np
import matplotlib.pyplot as plt
import jax.numpy as jnp
from src.data.toy.muller_brown import MullerBrownDataset, muller_brown_potential

# Create dataset with a few trajectories
dataset = MullerBrownDataset(size=5, seq_len=1000, dt=0.001, kT=15.0)

# Set up the grid for contour plot, strictly clipped to prevent blow up
x = np.linspace(-1.5, 1.2, 200)
y = np.linspace(-0.5, 2.0, 200)
X, Y = np.meshgrid(x, y)

# Compute potential over the grid
Z = np.zeros_like(X)
for i in range(X.shape[0]):
    for j in range(X.shape[1]):
        Z[i, j] = muller_brown_potential(jnp.array([X[i, j], Y[i, j]]))

# Clip Z to prevent extreme values from dominating the color map
Z_clipped = np.clip(Z, -200, 200)

plt.figure(figsize=(10, 8))
contour = plt.contourf(X, Y, Z_clipped, levels=50, cmap='viridis')
plt.colorbar(contour, label='Potential')

for i in range(len(dataset)):
    traj = dataset[i]['x_raw'].numpy()
    plt.plot(traj[:, 0], traj[:, 1], alpha=0.6, linewidth=1.5, label=f'Trajectory {i+1}')

plt.xlim([-1.5, 1.2])
plt.ylim([-0.5, 2.0])
plt.title('Langevin Dynamics on Müller-Brown Potential')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```
